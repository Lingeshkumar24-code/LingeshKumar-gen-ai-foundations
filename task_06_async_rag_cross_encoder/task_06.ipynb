# Task 6: High-Throughput Asynchronous RAG Pipeline with Cross-Encoder Reranking

## Dual-Stage Architecture
Bi-encoder candidate retrieval + Cross-encoder precision reranking.


```python
import asyncio

# Asynchronous RAG service orchestrating candidate retrieval and cross-encoder score reranking
class AsyncRAGPipeline:
    def __init__(self):
        self.corpus = [
            "Scaled dot-product attention computes QK^T scaled by sqrt(d_k).",
            "Bi-encoders embed queries and documents into dense vector spaces for fast FAISS candidate search.",
            "Cross-encoders process query and passage jointly through self-attention to produce accurate relevance scores.",
            "FastAPI uses asyncio to handle concurrent client queries under strict latency thresholds.",
            "Retrieval-Augmented Generation combines factual document retrieval with generative language models."
        ]
        
    async def retrieve_candidates(self, query: str):
        await asyncio.sleep(0.01) # Simulate async vector lookup
        return self.corpus
        
    async def rerank(self, query: str, candidates: list):
        await asyncio.sleep(0.01) # Simulate async cross-encoder scoring
        scores = [len(set(query.lower().split()) & set(c.lower().split())) for c in candidates]
        ranked = sorted(zip(scores, candidates), reverse=True)
        return ranked
        
    async def generate(self, query: str):
        candidates = await self.retrieve_candidates(query)
        ranked_results = await self.rerank(query, candidates)
        best_doc = ranked_results[0][1]
        return best_doc, ranked_results
```

```python
# Execute async query through dual-stage RAG pipeline
async def main():
    rag = AsyncRAGPipeline()
    query = "Why are cross-encoders used in dual-stage RAG?"
    best_context, all_ranked = await rag.generate(query)
    
    print("User Query:", f"'{query}'
")
    print("Cross-Encoder Candidate Re-scoring:")
    for rank, (score, doc) in enumerate(all_ranked[:3], 1):
        print(f"  Rank {rank} (Score: {score}): '{doc}'")
    print(f"
Selected Context for LLM Synthesis: '{best_context}'")

await main()
```
