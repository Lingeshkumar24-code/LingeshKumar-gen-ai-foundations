# Task 13: Direct Preference Optimization (DPO) and Pairwise Reward Alignment

## DPO Loss Formula
$$L_{DPO}(\theta) = - \log \sigma \left( \beta \log \frac{\pi_\theta(y_w|x)}{\pi_{ref}(y_w|x)} - \beta \log \frac{\pi_\theta(y_l|x)}{\pi_{ref}(y_l|x)} \right)$$


```python
import torch
import torch.nn.functional as F

# Direct Preference Optimization (DPO) pairwise loss calculation
def dpo_loss(policy_win_logps, policy_lose_logps, ref_win_logps, ref_lose_logps, beta=0.1):
    pi_logratios = policy_win_logps - policy_lose_logps
    ref_logratios = ref_win_logps - ref_lose_logps
    return -F.logsigmoid(beta * (pi_logratios - ref_logratios)).mean()
```

```python
# Compute DPO loss for chosen (win) and rejected (lose) completions
p_win, p_lose = torch.tensor([-1.2, -0.8]), torch.tensor([-2.5, -3.1])
r_win, r_lose = torch.tensor([-1.5, -1.0]), torch.tensor([-2.2, -2.8])

loss = dpo_loss(p_win, p_lose, r_win, r_lose)
print("DPO Pairwise Loss:", loss.item())
print("[SUCCESS] DPO pairwise loss verified.")
```
