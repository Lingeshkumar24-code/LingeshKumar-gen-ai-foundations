# Task 13: Direct Preference Optimization (DPO) and Pairwise Reward Alignment

## DPO Loss Formula
$$L_{DPO}(\theta) = - \log \sigma \left( \beta \log \frac{\pi_\theta(y_w|x)}{\pi_{ref}(y_w|x)} - \beta \log \frac{\pi_\theta(y_l|x)}{\pi_{ref}(y_l|x)} \right)$$


```python
import torch
import torch.nn.functional as F

# DPO loss computing relative log odds ratio margin between chosen and rejected completions
def dpo_loss(policy_win_logps, policy_lose_logps, ref_win_logps, ref_lose_logps, beta=0.1):
    pi_logratios = policy_win_logps - policy_lose_logps
    ref_logratios = ref_win_logps - ref_lose_logps
    logits = pi_logratios - ref_logratios
    loss = -F.logsigmoid(beta * logits)
    return loss.mean(), logits.detach()
```

```python
# Process preference dataset sample pairs (chosen y_w vs rejected y_l)
prompts = [
    "How do I handle errors in Python?",
    "What is the best way to store passwords?"
]

policy_win = torch.tensor([-1.2, -0.8])
policy_lose = torch.tensor([-2.5, -3.1])
ref_win = torch.tensor([-1.5, -1.0])
ref_lose = torch.tensor([-2.2, -2.8])

loss_val, margins = dpo_loss(policy_win, policy_lose, ref_win, ref_lose, beta=0.1)

print("DPO Preference Alignment Sample Logs:")
for i, prompt in enumerate(prompts):
    print(f"  Prompt #{i+1}: '{prompt}'")
    print(f"    Preference Margin (y_w vs y_l): {margins[i].item():+.4f}")

print(f"
Average Pairwise DPO Loss: {loss_val.item():.4f}")
```
