# Task 12: Knowledge Distillation of Dual-Encoder Vision-Language Models

## Composite Loss
$$L_{total} = \alpha L_{cosine} + (1 - \alpha) T^2 D_{KL}$$


```python
import torch
import torch.nn.functional as F

# Composite distillation loss combining soft-label KL divergence and cosine distance
def distillation_loss(student_logits, teacher_logits, temperature=2.0, alpha=0.5):
    kl = F.kl_div(
        F.log_softmax(student_logits / temperature, dim=-1),
        F.softmax(teacher_logits / temperature, dim=-1),
        reduction='batchmean'
    ) * (temperature ** 2)
    cosine = 1.0 - F.cosine_similarity(student_logits, teacher_logits).mean()
    return alpha * cosine + (1.0 - alpha) * kl, kl.item(), cosine.item()
```

```python
# Compute distillation loss for mini dataset of image-text embeddings
teacher_features = torch.randn(4, 512) # 512-dim heavy CLIP teacher
student_features = torch.randn(4, 512) # 512-dim lightweight student projection

total_loss, kl_val, cos_val = distillation_loss(student_features, teacher_features, temperature=2.0, alpha=0.5)

print("Teacher Embedding Dimension: 512")
print("Student Embedding Dimension: 512")
print(f"Temperature T: 2.0 | Loss Weight Alpha: 0.5
")
print(f"Soft-Label KL Divergence Loss: {kl_val:.4f}")
print(f"Cosine Distance Loss:         {cos_val:.4f}")
print(f"Total Composite Distillation Loss: {total_loss.item():.4f}")
```
