# Task 12: Knowledge Distillation of Dual-Encoder Vision-Language Models

## Composite Loss
$$L_{total} = \alpha L_{cosine} + (1 - \alpha) T^2 D_{KL}$$


```python
import torch
import torch.nn.functional as F

# Composite loss combining temperature-scaled KL divergence and cosine distance
def distillation_loss(student_logits, teacher_logits, temperature=2.0, alpha=0.5):
    kl = F.kl_div(
        F.log_softmax(student_logits / temperature, dim=-1),
        F.softmax(teacher_logits / temperature, dim=-1),
        reduction='batchmean'
    ) * (temperature ** 2)
    cosine = 1.0 - F.cosine_similarity(student_logits, teacher_logits).mean()
    return alpha * cosine + (1.0 - alpha) * kl
```

```python
# Calculate composite distillation loss between Teacher and Student embeddings
teacher_out = torch.randn(4, 512)
student_out = torch.randn(4, 512)
loss = distillation_loss(student_out, teacher_out)

print("Distillation Composite Loss:", loss.item())
print("[SUCCESS] Vision-language distillation loss verified.")
```
